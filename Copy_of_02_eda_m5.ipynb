```python
df['bmi_category'] = pd.cut(df['bmi'], bins=[0, 18.5, 25, 30, 100], labels=['Underweight', 'Normal', 'Overweight', 'Obese'], right=False)
sns.barplot(x='bmi_category', y='charges', hue='smoker', data=df, errorbar=None)
plt.title('Average Charges by BMI Category')
plt.show()
```

```python
sns.barplot(x='region', y='charges', data=df)
plt.title('Average Charges by Region')
plt.show()
```

```python
# ## Smoking combined with BMI
# - Non-smokers pay about the same whatever their BMI category.
# - Smokers with BMI 30 or more average about 41,558.
# - Smokers with BMI below 30 average about 21,363.
# - Smoking with a high BMI is the most expensive group.
```
