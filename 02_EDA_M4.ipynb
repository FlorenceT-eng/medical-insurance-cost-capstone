```python
#load the cleaned data
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/insurance_clean.csv')
print(df.shape)
df.head()
```

```python
#age vs charges, coloured by smoker
sns.scatterplot(x='age', y='charges', hue='smoker', data=df)
plt.title('Age vs Insurance Charges')
plt.show()
print("Charges increase with age for everyone, and smokers pay far more at every age.")
```

```python
#average charges by age band
df['age_band'] = pd.cut(df['age'], bins=[17, 30, 40, 50, 65],
                        labels=['18-30', '31-40', '41-50', '51-64'])
sns.barplot(x='age_band', y='charges', data=df)
plt.title('Average Charges by Age Band')
plt.show()
print("Average charges roughly double from the youngest to the oldest age band.")
```

```python
#charges by number of children
sns.barplot(x='children', y='charges', data=df)
plt.title('Average Charges by Number of Children')
plt.show()
print("Number of children has only a small effect on charges. The 4 and 5 children groups are too small to trust")
```
