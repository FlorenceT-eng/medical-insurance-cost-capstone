```python
# M6:Medical insurance cost modelling

# Step 1: import necessary libaries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/insurance_clean.csv")
print(df.shape)
df.head()
```

```python
#Check dataset information
df.info()

#Check for missing values
print("Missing values:")

print(df.isnull().sum())

# Check for duplicate rows
print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
# Seperate features and target

x = df.drop('charges', axis=1)
y = df['charges']

print("Features shape:", x.shape)
print("Target shape:", y.shape)
```

**Splitting the data into 80% training and 20% testing, using ramdom_state=42**

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size = 0.2,
    random_state = 42
)

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)
```

```python
# Import necessary libaries
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create baseline model predictions using the traning mean
baseline_predictions = np.full(len(y_test), y_train.mean())

# Evaluate the model
baseline_mae = mean_absolute_error(y_test, baseline_predictions)
baseline_rmse = np.sqrt(mean_squared_error(y_test, baseline_predictions))
baseline_r2 = r2_score(y_test, baseline_predictions)

print("Baseline Model Results")
print("Mae:", baseline_mae)
print("RMSE:", baseline_rmse)
print("R² Score:", baseline_r2)
                                   
```

### Baseline Model

A baseline regression model was established using the mean of the training target values as the prediction for every observation in the test dataset.

The model was evaluated using Mean absolute Error (MAE), Root Mean Square Error (RMSE), and R-squared (R²) to provide the benchmark for comparing the performanceof more advanced models.
